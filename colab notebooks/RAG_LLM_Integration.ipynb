```python
import json

input_path = "/content/drive/MyDrive/storage/documents.json"
output_path = "/content/drive/MyDrive/RAG_data_collector_module/lora_train.jsonl"

with open(input_path, "r", encoding="utf-8") as f:
    data = json.load(f)

converted = []
for entry in data:
    page_text = entry.get("page_content", "").strip()
    if page_text:
        new_text = {
            "text": f"Instruction: Summarize or explain this content.\nContent: {page_text}\nAnswer:"
        }
        converted.append(new_text)

with open(output_path, "w", encoding="utf-8") as f_out:
    for item in converted:
        f_out.write(json.dumps(item) + "\n")

print(f"Converted {len(converted)} entries to LoRA SFT format.")
```

```python
from datasets import Dataset
import json
from transformers import AutoTokenizer

# Load dataset from JSONL
jsonl_path = "/content/drive/MyDrive/RAG_data_collector_module/lora_train.jsonl"
with open(jsonl_path, "r", encoding="utf-8") as f:
    lines = [json.loads(line) for line in f]
dataset = Dataset.from_list(lines)

# Load llama tokenizer
tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-3.2-1B")

# Add pad token
tokenizer.pad_token = tokenizer.eos_token

# Tokenization function
def tokenize(example):
    return tokenizer(example["text"], padding="max_length", truncation=True, max_length=512)

# Apply tokenization
tokenized_dataset = dataset.map(tokenize, batched=True)
tokenized_dataset = tokenized_dataset.remove_columns(["text"])

print("Tokenization complete. Sample:")
print(tokenized_dataset[0])
```

```python
from transformers import AutoModelForCausalLM
from peft import get_peft_model, LoraConfig, TaskType

# Load pre-trained base model
model = AutoModelForCausalLM.from_pretrained(
    "meta-llama/Llama-3.2-1B",
    device_map="auto",
    torch_dtype="auto",
    trust_remote_code=True
)

# LoRA Configuration
lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],  # Customize based on model
    lora_dropout=0.05,
    bias="none",
    task_type=TaskType.CAUSAL_LM
)

# Wrap model with PEFT
lora_model = get_peft_model(model, lora_config)
lora_model.print_trainable_parameters()
```

```python
from transformers import TrainingArguments, Trainer, DataCollatorForLanguageModeling
import os

os.environ["CUDA_LAUNCH_BLOCKING"] = "1"

lora_model.config.pad_token_id = tokenizer.pad_token_id

training_args = TrainingArguments(
    output_dir="/content/drive/MyDrive/lora_llama_finetuned",
    per_device_train_batch_size=2,
    gradient_accumulation_steps=4,
    num_train_epochs=2,
    learning_rate=2e-4,
    fp16=True,
    logging_dir="./logs",
    save_strategy="epoch",
    save_total_limit=1,
    logging_steps=10,
    report_to="none"
)

data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

trainer = Trainer(
    model=lora_model,
    args=training_args,
    train_dataset=tokenized_dataset,
    tokenizer=tokenizer,
    data_collator=data_collator
)

trainer.train()
```

```python
# !pip uninstall -y transformers huggingface_hub safetensors
# !pip install torch==2.1.0
# !pip install transformers==4.36.2
# !pip install huggingface_hub==0.20.3
# !pip install accelerate==0.25.0 bitsandbytes

from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain.prompts import PromptTemplate
from langchain.chains import RetrievalQA
from dotenv import load_dotenv
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
from langchain_community.llms import HuggingFacePipeline

import os
import sys
import torch

from huggingface_hub import login
login(token="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")

from google.colab import drive
drive.mount('/content/drive')

PROJECT_ROOT = "/content/drive/MyDrive"
os.chdir(PROJECT_ROOT)

if PROJECT_ROOT not in sys.path:
    sys.path.append(PROJECT_ROOT)

load_dotenv(dotenv_path=os.path.join(PROJECT_ROOT, ".env"))

def load_vector_db():
    embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
    vectorstore = Chroma(
        persist_directory=os.path.join(PROJECT_ROOT, "vector_store"),
        embedding_function=embeddings
    )
    return vectorstore

# Load Local LLM from HuggingFace (NO cloud call)
def load_llm():
    finetuned_model_path = "/content/drive/MyDrive/lora_llama_finetuned"

    tokenizer = AutoTokenizer.from_pretrained(finetuned_model_path, trust_remote_code=True)
    model = AutoModelForCausalLM.from_pretrained(
        finetuned_model_path,
        torch_dtype=torch.float16,
        device_map="auto",
        trust_remote_code=True
    )

    pipe = pipeline(
        "text-generation",
        model=model,
        tokenizer=tokenizer,
        max_new_tokens=512,
        temperature=1.0,
        repetition_penalty=1.1,
        do_sample=False
    )

    return HuggingFacePipeline(pipeline=pipe)

# RAG Pipeline
def build_rag_chain():
    retriever = load_vector_db().as_retriever(search_kwargs={"k": 5})
    llm = load_llm()

    template = """You are a helpful assistant. Use the following context to answer the question.
    If you don't know the answer, just say you don't know. Don't repeat any answer again. Each bullet should contain unique answer, and if there is no unique answer left just stop.

    Context:
    {context}

    Question:
    {question}

    Answer:"""

    prompt = PromptTemplate.from_template(template)

    rag_chain = RetrievalQA.from_chain_type(
        llm=llm,
        retriever=retriever,
        chain_type="stuff",
        chain_type_kwargs={"prompt": prompt},
        return_source_documents=True
    )
    return rag_chain

if __name__ == "__main__":
    rag = build_rag_chain()
    while True:
        question = input("\n❓ Enter your question (or type 'exit'): ").strip()
        if question.lower() == 'exit':
            break

        result = rag.invoke({"query": question})
        print("\n🔍 Answer:", result["result"])

        print("\n📚 Sources:")
        for doc in result["source_documents"]:
            print(f" - {doc.metadata.get('source', 'Unknown source')}")
```
